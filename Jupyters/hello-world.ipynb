```python
# ==========================================
# SIMPLE NEXT-WORD PREDICTOR
# ==========================================

# Training-like examples
sentences = [
    "I am a software engineer",
    "I am a software developer",
    "I am a software architect",
    "I am a software engineer",
    "I am a software engineer"
]

# Count what words appear after "I am a software"
next_words = {}

for sentence in sentences:

    words = sentence.split()

    # Find the phrase
    for i in range(len(words) - 1):

        current_word = words[i]
        next_word = words[i + 1]

        if current_word not in next_words:
            next_words[current_word] = {}

        if next_word not in next_words[current_word]:
            next_words[current_word][next_word] = 0

        next_words[current_word][next_word] += 1


# Show learned word relationships
print("Word relationships:")
print(next_words)


# Predict the next word after "software"
word = "software"

if word in next_words:

    predictions = next_words[word]

    predicted_word = max(
        predictions,
        key=predictions.get
    )

    print("\nInput word:", word)
    print("Predicted next word:", predicted_word)

else:

    print("No prediction available.")
```
